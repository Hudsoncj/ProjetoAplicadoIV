# **Referencial Teórico**


**5 REFERENCIAL TEÓRICO**

**5.1 Séries Temporais e Previsão de Carga Elétrica**

A previsão da carga elétrica é fundamental para o planejamento e a operação do Sistema Interligado Nacional (SIN), permitindo estimar a demanda e apoiar a utilização eficiente dos recursos energéticos. Segundo Morettin e Toloi (2018), séries temporais são conjuntos de observações ordenadas cronologicamente, que podem apresentar tendência, sazonalidade, ciclos e componentes aleatórios.

No setor elétrico, a demanda apresenta dependência temporal e pode sofrer influência de fatores como temperatura, calendário, atividade econômica e hábitos de consumo. A identificação desses padrões permite compreender o comportamento histórico da carga e selecionar técnicas adequadas para sua previsão (Taylor; McSharry, 2007).

<br>A análise de séries temporais envolve conceitos fundamentais, como:

- **Tendência:** comportamento de crescimento, redução ou estabilidade da série ao longo do tempo.
- **Sazonalidade:** padrões que se repetem em intervalos regulares, como variações semanais e anuais.
- **Ciclos:** oscilações recorrentes que não apresentam necessariamente periodicidade fixa.
- **Ruído:** variações irregulares não explicadas pelos componentes identificados.
- **Estacionariedade:** propriedade relacionada à estabilidade de características estatísticas da série, como média, variância e estrutura de autocovariância.

<br>Esses componentes podem ser investigados por meio de análise exploratória, autocorrelação (ACF), autocorrelação parcial (PACF), testes de estacionariedade e técnicas de decomposição temporal, conforme discutido por Morettin e Toloi (2018) e Hyndman e Athanasopoulos (2021).

Entre os modelos aplicáveis destacam-se o SARIMA, que representa dependências temporais e sazonais; o SARIMAX, que permite incorporar variáveis externas; e os modelos estruturais de espaço de estados, que representam componentes não observáveis e podem utilizar o Filtro de Kalman para estimação (Harvey, 1989; Hyndman; Athanasopoulos, 2021).

<br>**5.2 Trabalhos Correlatos**

**Taylor e McSharry (2007)** compararam métodos de previsão de carga elétrica de curto prazo utilizando dados de dez países europeus. Foram avaliados modelos ARIMA, autorregressivos periódicos, suavização exponencial e análise de componentes principais (PCA). Os resultados indicaram bom desempenho dos modelos ARIMA e PCA, enquanto Holt-Winters com dupla sazonalidade apresentou os melhores resultados de forma consistente.

O estudo demonstra a importância de considerar múltiplos padrões sazonais e fundamenta a utilização de modelos estatísticos como alternativas iniciais para previsão de carga elétrica. Sua principal limitação para este projeto é a utilização de dados intradiários, enquanto a base selecionada do ONS apresenta frequência diária.

<br>**Hong e Fan (2016)** realizaram uma revisão sobre previsão probabilística de carga elétrica, discutindo métodos estatísticos, aplicações e procedimentos de avaliação. Os autores destacam a importância de representar a incerteza associada às previsões, especialmente em decisões relacionadas ao planejamento e à operação de sistemas energéticos.

O trabalho contribui para a definição de critérios de avaliação e para a compreensão das limitações de previsões exclusivamente pontuais. Entretanto, por ser uma revisão metodológica, não estabelece um único modelo como superior para todos os conjuntos de dados.

<br>**Guo et al. (2021)** investigaram métodos de aprendizado de máquina aplicados à previsão de carga elétrica de curto prazo, incluindo Support Vector Machine (SVM), Random Forest (RF) e redes neurais Long Short-Term Memory (LSTM).

Os autores analisaram vantagens e limitações dessas técnicas e propuseram uma abordagem combinada de previsão, associada ao pré-processamento dos dados. Os experimentos com dados reais indicaram ganhos de precisão da abordagem combinada em relação às alternativas avaliadas.

O estudo demonstra que métodos de aprendizado de máquina podem representar relações complexas e não lineares. Entretanto, sua aplicação pode exigir maior esforço de treinamento, seleção de atributos, ajuste de parâmetros e recursos computacionais. Esses fatores são relevantes para a escolha de uma solução viável no Google Colab.

Os três trabalhos evidenciam que não existe uma técnica universalmente superior para previsão de carga elétrica. O desempenho depende das características dos dados, da frequência de amostragem, do horizonte de previsão e dos padrões temporais presentes nas observações.

<br>**5.3 Técnicas Propostas e Justificativa**

Considerando os trabalhos analisados e as características da carga diária do ONS, propõe-se comparar um modelo estatístico SARIMA/SARIMAX com um modelo estrutural de espaço de estados, utilizando também uma previsão sazonal ingênua como referência (*baseline*).

O SARIMA permite modelar dependências temporais e padrões sazonais, enquanto o SARIMAX possibilita incorporar variáveis exógenas, desde que estejam disponíveis no momento da previsão (Hyndman; Athanasopoulos, 2021).

Os modelos estruturais de espaço de estados permitem representar componentes não observáveis, como nível, tendência e sazonalidade. A estimação desses componentes pode ser realizada por meio do Filtro de Kalman (Harvey, 1989).

A utilização de uma previsão sazonal ingênua permite verificar se os modelos mais complexos produzem ganhos efetivos de desempenho em relação a uma abordagem simples, conforme os princípios de avaliação de previsões apresentados por Hyndman e Athanasopoulos (2021).

<br>**Comparação das Técnicas Candidatas**<br><br>

| Técnica | Vantagens | Limitações |
|---|---|---|
| Baseline sazonal | Simples, rápido e de fácil interpretação | Não representa adequadamente mudanças de tendência ou padrões complexos |
| SARIMA | Modela autocorrelação e sazonalidade de forma estatisticamente estruturada | Pode apresentar dificuldades com relações não lineares e múltiplas sazonalidades |
| SARIMAX | Permite incorporar variáveis externas, como temperatura e calendário | Depende da disponibilidade, qualidade e previsibilidade das variáveis externas |
| Modelo Estrutural de Espaço de Estados | Permite representar componentes estruturais e atualizá-los recursivamente | Exige especificação adequada dos componentes e estimação dos parâmetros |

<br>A comparação considera que modelos SARIMA também podem ser representados matematicamente em espaço de estados. Portanto, para garantir uma avaliação efetiva entre abordagens distintas, será utilizado um modelo estrutural explicitamente definido por componentes de nível, tendência e sazonalidade.

A escolha final considerará a precisão das previsões, estabilidade, interpretabilidade e complexidade computacional. Serão utilizadas métricas como MAE, RMSE e MAPE, permitindo comparar os resultados por subsistema regional.

# **EDA e Pré-processamento dos dados**


````python
# Baixar e consolidar os dados


import pandas as pd
from io import BytesIO
import requests

# Período definido no projeto
anos = range(2015, 2026)

# Fonte oficial do ONS
url_base = (
    "https://ons-aws-prod-opendata.s3.amazonaws.com/"
    "dataset/carga_energia_di/CARGA_ENERGIA_{}.csv"
)

dataframes = []

for ano in anos:
    url = url_base.format(ano)
    print(f"Baixando dados de {ano}...")

    resposta = requests.get(url, timeout=60)
    resposta.raise_for_status()

    df_ano = pd.read_csv(
        BytesIO(resposta.content),
        sep=";",
        encoding="utf-8"
    )

    dataframes.append(df_ano)

# Consolidar os 11 anos em um DataFrame
df = pd.concat(dataframes, ignore_index=True)

# Conversões iniciais
df["din_instante"] = pd.to_datetime(
    df["din_instante"], errors="coerce"
)

# Ajustado para o nome correto da coluna de carga: val_cargaenergiamwmed
df["val_cargaenergiamwmed"] = pd.to_numeric(
    df["val_cargaenergiamwmed"], errors="coerce"
)

# Ordenação cronológica
df = df.sort_values(
    ["id_subsistema", "din_instante"]
).reset_index(drop=True)

print(f"\nTotal de registros: {len(df):,}")
print(f"Período: {df['din_instante'].min()} até {df['din_instante'].max()}")

display(df.head())

# Salvar arquivo consolidado
df.to_csv(
    "carga_energia_2015_2025.csv",
    index=False
)
````

````python

print("Colunas encontradas:")
print(df.columns.tolist())

print("\nPrimeiros registros:")
display(df.head(5))

print("\nDimensões:", df.shape)
````

````python

# Quantidade de registros por ano e subsistema
df["ano"] = df["din_instante"].dt.year

tabela = pd.crosstab(
    df["ano"],
    df["id_subsistema"]
)

display(tabela)

# Conferir continuidade diária por subsistema
datas_esperadas = pd.date_range(
    "2015-01-01", "2025-12-31", freq="D"
)

for subsistema, grupo in df.groupby("id_subsistema"):
    datas_encontradas = pd.DatetimeIndex(
        grupo["din_instante"].dropna().unique()
    )

    faltantes = datas_esperadas.difference(
        datas_encontradas
    )

    print(
        f"Subsistema {subsistema}: "
        f"{len(faltantes)} dias faltantes"
    )
````

````python
# Verificação de valores nulos, duplicações ou valores de carga inválidos

print("=== VALIDAÇÃO FINAL ===")

print("\nValores nulos:")
print(df.isnull().sum())

print("\nRegistros duplicados:")
print(df.duplicated(
    subset=["id_subsistema", "din_instante"]
).sum())

print("\nValores de carga negativos ou iguais a zero:")
print((df["val_cargaenergiamwmed"] <= 0).sum())

print("\nEstatísticas da carga elétrica por subsistema:")
display(
    df.groupby("id_subsistema")["val_cargaenergiamwmed"]
      .describe()
      .round(2)
)
````

````python
# Identificar registros com carga nula antes de realizar a analise estatística

nulos = df[df["val_cargaenergiamwmed"].isna()]

print("=== REGISTROS COM CARGA NULA ===")
display(nulos)

# Verificar os registros anteriores e posteriores
for idx in nulos.index:
    subsistema = df.loc[idx, "id_subsistema"]
    data = df.loc[idx, "din_instante"]

    print(f"\nSubsistema: {subsistema} | Data: {data}")

    vizinhos = df[
        (df["id_subsistema"] == subsistema) &
        (df["din_instante"].between(
            data - pd.Timedelta(days=3),
            data + pd.Timedelta(days=3)
        ))
    ]

    display(vizinhos[
        ["din_instante", "val_cargaenergiamwmed"]
    ])
````

````python
# Valores ausentes entre dois dias com registros válidos.
# Vamos utilizar interpolação linear temporal, separadamente, por subsistema
# A idéia é preservar a continuidade das séries e alterar apenas 4 observações (0,025% da base)

import pandas as pd

# Preservar os dados originais
df_tratado = df.copy()

coluna = "val_cargaenergiamwmed"

# Garantir tipos corretos
df_tratado["din_instante"] = pd.to_datetime(
    df_tratado["din_instante"]
)

df_tratado[coluna] = pd.to_numeric(
    df_tratado[coluna], errors="coerce"
)

# Ordenar por subsistema e data
df_tratado = df_tratado.sort_values(
    ["id_subsistema", "din_instante"]
).reset_index(drop=True)

# Registrar os valores originalmente nulos
mascara_nulos = df_tratado[coluna].isna()

# Interpolar separadamente por subsistema
def interpolar_grupo(grupo):
    grupo = grupo.copy()
    grupo = grupo.set_index("din_instante")

    grupo[coluna] = grupo[coluna].interpolate(
        method="time",
        limit_area="inside"
    )

    return grupo.reset_index()

df_tratado = pd.concat(
    [
        interpolar_grupo(grupo)
        for _, grupo in df_tratado.groupby(
            "id_subsistema", sort=False
        )
    ],
    ignore_index=True
)

# Conferir os quatro registros corrigidos
print("=== VALORES INTERPOLADOS ===")

display(
    df_tratado.loc[
        mascara_nulos,
        [
            "id_subsistema",
            "din_instante",
            coluna
        ]
    ]
)

print("\nValores nulos restantes:")
print(df_tratado[coluna].isna().sum())

# Salvar a base tratada
df_tratado.to_csv(
    "carga_energia_2015_2025_tratada.csv",
    index=False
)

print("\nBase tratada salva com sucesso!")
````

<br>**Alterações Metodológicas na Composição da Carga Elétrica**

A análise da série histórica deve considerar alterações na metodologia de contabilização da carga elétrica disponibilizada pelo Operador Nacional do Sistema Elétrico (ONS).

- Até fevereiro de 2021, os dados diários representavam a carga atendida por usinas despachadas ou programadas pelo ONS.
- A partir de março de 2021, passaram a incorporar também a previsão de geração de usinas não despachadas diretamente pelo operador.
- Em 29 de abril de 2023, ocorreu outra alteração relevante, com a inclusão do valor estimado da micro e minigeração distribuída (MMGD), principalmente associada à geração solar instalada em residências, comércios e indústrias.

A MMGD permite que parte da energia consumida seja produzida localmente, reduzindo a necessidade de fornecimento pela rede elétrica. Sua incorporação aos dados do ONS busca representar de maneira mais completa a carga global do sistema.

Entretanto, essa mudança metodológica pode provocar alterações no nível, na tendência e na sazonalidade das séries temporais, sem que necessariamente representem mudanças proporcionais no consumo real de energia.

Dessa forma, a análise exploratória considerará as alterações de 2021 e 2023 como possíveis pontos de mudança estrutural, investigando seus efeitos sobre os quatro subsistemas do SIN.

Essas informações serão consideradas na interpretação dos gráficos, nos testes estatísticos e na avaliação dos modelos preditivos, evitando atribuir exclusivamente ao comportamento do consumo variações decorrentes de alterações metodológicas.

<br>**8.3 Decomposição das Séries Temporais — MSTL**

A análise exploratória identificou padrões sazonais semanais e anuais na carga diária de energia elétrica dos quatro subsistemas do SIN.

Para investigar esses componentes simultaneamente, será utilizada a técnica MSTL (Multiple Seasonal-Trend decomposition using Loess), uma extensão do método STL que permite decompor séries temporais com múltiplas sazonalidades.

A decomposição aditiva representa a série observada pela combinação de tendência, componentes sazonais e resíduos:

$$
Y_t = T_t + S_{7,t} + S_{365,t} + R_t
$$

Em que:

- **Y_t:** carga elétrica diária observada.
- **T_t:** tendência de longo prazo.
- **S_{7,t}:** componente sazonal semanal, com período de 7 dias.
- **S_{365,t}:** componente sazonal anual, com período aproximado de 365 dias.
- **R_t:** componente residual, não explicado pela tendência e pelas sazonalidades.

A aplicação será realizada separadamente para os subsistemas Norte, Nordeste, Sul e Sudeste/Centro-Oeste, utilizando a base `df_modelagem`, previamente tratada e validada.

Também serão consideradas as alterações metodológicas na contabilização da carga pelo ONS em 2021 e 2023, pois podem influenciar a tendência e os componentes identificados.

A decomposição será utilizada para fins de análise exploratória. Durante a modelagem preditiva, os procedimentos de treinamento e validação deverão evitar a utilização de informações futuras.<br><br>


````python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels

from statsmodels.tsa.seasonal import MSTL

print("Versão Statsmodels:", statsmodels.__version__)

# Base preparada anteriormente
dados = df_modelagem.copy()

coluna = "val_cargaenergiamwmed"

subsistemas = {
    "N": "Norte",
    "NE": "Nordeste",
    "S": "Sul",
    "SE": "Sudeste/Centro-Oeste"
}

# Armazenar os resultados da decomposição
resultados_mstl = {}

for sigla, nome in subsistemas.items():

    print(f"\n=== DECOMPOSIÇÃO MSTL: {nome} ===")

    # Selecionar subsistema e organizar série temporal
    serie = (
        dados[dados["id_subsistema"] == sigla]
        .sort_values("din_instante")
        .set_index("din_instante")[coluna]
        .asfreq("D")
    )

    # Garantir integridade dos dados
    if serie.isna().any():
        raise ValueError(
            f"Existem valores ausentes no subsistema {sigla}"
        )

    # Aplicar decomposição com sazonalidades
    # semanal (7 dias) e anual (365 dias)
    modelo = MSTL(
        serie,
        periods=(7, 365),
        windows=(11, 15),
        iterate=2,
        stl_kwargs={
            "robust": False,
            "seasonal_jump": 7,
            "trend_jump": 7,
            "low_pass_jump": 7
        }
    )

    resultado = modelo.fit()

    # Organizar componentes
    componentes = pd.DataFrame({
        "observado": serie,
        "tendencia": resultado.trend,
        "semanal": resultado.seasonal["seasonal_7"],
        "anual": resultado.seasonal["seasonal_365"],
        "residuo": resultado.resid
    })

    resultados_mstl[sigla] = componentes

    # Gráficos da decomposição
    fig, axes = plt.subplots(
        5, 1,
        figsize=(15, 12),
        sharex=True
    )

    componentes["observado"].plot(
        ax=axes[0], title="Série original"
    )

    componentes["tendencia"].plot(
        ax=axes[1], title="Tendência"
    )

    componentes["semanal"].plot(
        ax=axes[2], title="Sazonalidade semanal"
    )

    componentes["anual"].plot(
        ax=axes[3], title="Sazonalidade anual"
    )

    componentes["residuo"].plot(
        ax=axes[4], title="Resíduos"
    )

    for ax in axes:
        ax.grid(alpha=0.3)
        ax.set_ylabel("MWmédio")

    axes[-1].set_xlabel("Ano")

    fig.suptitle(
        f"Decomposição MSTL — {nome} (2015–2025)",
        fontsize=15
    )

    plt.tight_layout()
    plt.show()

print("\nDecomposição concluída.")
````

````python

print("=== VALIDAÇÃO DA DECOMPOSIÇÃO ===")

for sigla, componentes in resultados_mstl.items():

    reconstruida = (
        componentes["tendencia"]
        + componentes["semanal"]
        + componentes["anual"]
        + componentes["residuo"]
    )

    erro = (
        componentes["observado"] - reconstruida
    ).abs().max()

    print(
        f"{sigla}: erro máximo de reconstrução = "
        f"{erro:.10f} MWmédio"
    )
````

````python
# Medir a força da tendência e das sazonalidades


import numpy as np
import pandas as pd

print("=== FORÇA DOS COMPONENTES MSTL ===")

resultados = []

for sigla, comp in resultados_mstl.items():

    residuo = comp["residuo"]
    tendencia = comp["tendencia"]
    semanal = comp["semanal"]
    anual = comp["anual"]

    # Indicadores de força dos componentes
    var_residuo = residuo.var()

    forca_tendencia = max(
        0, 1 - var_residuo /
        (residuo + tendencia).var()
    )

    forca_semanal = max(
        0, 1 - var_residuo /
        (residuo + semanal).var()
    )

    forca_anual = max(
        0, 1 - var_residuo /
        (residuo + anual).var()
    )

    resultados.append({
        "Subsistema": sigla,
        "Força Tendência": forca_tendencia,
        "Força Semanal": forca_semanal,
        "Força Anual": forca_anual,
        "Desvio Resíduo": residuo.std()
    })

df_forca_mstl = pd.DataFrame(resultados)

display(
    df_forca_mstl.round(3)
)
````

<br>**8.4 Análise de Autocorrelação e Estacionariedade**

<br>Após a identificação dos componentes de tendência e sazonalidade por meio da decomposição MSTL, serão analisadas as propriedades estatísticas das séries temporais.

A Função de Autocorrelação (ACF) mede a relação entre observações separadas por diferentes defasagens temporais (*lags*), enquanto a Função de Autocorrelação Parcial (PACF) avalia essa relação controlando a influência das defasagens intermediárias.

Para a carga diária de energia elétrica, serão investigadas especialmente as defasagens de 7, 14, 21 e 28 dias, buscando identificar a persistência da sazonalidade semanal.

Também serão aplicados os testes Augmented Dickey-Fuller (ADF) e Kwiatkowski-Phillips-Schmidt-Shin (KPSS) para investigar a estacionariedade das séries.

A análise será realizada inicialmente sobre os dados originais e, posteriormente, sobre as séries com primeira diferenciação, permitindo avaliar alterações na estrutura de autocorrelação e na estabilidade estatística.

Os resultados serão utilizados para orientar a configuração dos modelos estatísticos de previsão.<br><br>


````python

import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

# Base final tratada
dados = df_modelagem.copy()

coluna = "val_cargaenergiamwmed"

dados["din_instante"] = pd.to_datetime(
    dados["din_instante"]
)

subsistemas = {
    "N": "Norte",
    "NE": "Nordeste",
    "S": "Sul",
    "SE": "Sudeste/Centro-Oeste"
}

# Armazenar séries para análises posteriores
series_temporais = {}

for sigla, nome in subsistemas.items():

    # Série original
    serie = (
        dados[dados["id_subsistema"] == sigla]
        .sort_values("din_instante")
        .set_index("din_instante")[coluna]
        .asfreq("D")
    )

    if serie.isna().any():
        raise ValueError(
            f"Existem valores ausentes no subsistema {sigla}"
        )

    # Primeira diferenciação
    serie_diff = serie.diff().dropna()

    series_temporais[sigla] = {
        "original": serie,
        "diferenciada": serie_diff
    }

    # Quatro gráficos por subsistema
    fig, axes = plt.subplots(
        2, 2,
        figsize=(14, 8)
    )

    # ACF da série original
    plot_acf(
        serie,
        lags=35,
        ax=axes[0, 0],
        title="ACF — Série Original"
    )

    # PACF da série original
    plot_pacf(
        serie,
        lags=35,
        method="ywm",
        ax=axes[0, 1],
        title="PACF — Série Original"
    )

    # ACF da série diferenciada
    plot_acf(
        serie_diff,
        lags=35,
        ax=axes[1, 0],
        title="ACF — Primeira Diferença"
    )

    # PACF da série diferenciada
    plot_pacf(
        serie_diff,
        lags=35,
        method="ywm",
        ax=axes[1, 1],
        title="PACF — Primeira Diferença"
    )

    for ax in axes.flat:
        for lag in [7, 14, 21, 28, 35]:
            ax.axvline(
                lag,
                color="gray",
                linestyle=":",
                linewidth=0.8,
                alpha=0.5
            )
        ax.set_xlabel("Defasagem (dias)")
        ax.grid(alpha=0.2)

    fig.suptitle(
        f"Autocorrelação — Subsistema {nome}",
        fontsize=15
    )

    plt.tight_layout()
    plt.show()
````

````python
# Testes ADF e KPSS


import pandas as pd
import warnings

from statsmodels.tsa.stattools import adfuller, kpss
from statsmodels.tools.sm_exceptions import InterpolationWarning

resultados_testes = []

for sigla, series in series_temporais.items():

    for tipo, serie in series.items():

        # Teste ADF

        adf = adfuller(
            serie,
            regression="c",
            autolag="AIC",
            result_object=False
        )

        # Teste KPSS
        # Avisos de limites da tabela de p-valores
        # serão tratados na interpretação
        with warnings.catch_warnings():
            warnings.simplefilter(
                "ignore", InterpolationWarning
            )


            kpss_resultado = kpss(
                serie,
                regression="c",
                nlags="auto",
                result_object=False
            )

        resultados_testes.append({
            "Subsistema": sigla,
            "Série": tipo,
            "ADF estatística": adf[0],
            "ADF p-valor": adf[1],
            "KPSS estatística": kpss_resultado[0],
            "KPSS p-valor": kpss_resultado[1]
        })

df_testes = pd.DataFrame(resultados_testes)

print("=== TESTES DE ESTACIONARIEDADE ===")
display(df_testes.round(4))
````


**8.4.1 Resultados da Análise de Autocorrelação e Estacionariedade**

<br>A análise das funções ACF e PACF demonstrou forte dependência temporal nas séries de carga elétrica dos quatro subsistemas do SIN.

Nas séries originais, foram identificadas autocorrelações elevadas e persistentes, associadas à presença de tendência e sazonalidade. Também foram observados picos recorrentes nas defasagens de 7, 14, 21, 28 e 35 dias, evidenciando a importância do padrão semanal.

Após a primeira diferenciação, houve redução da autocorrelação em diversas defasagens. Entretanto, os picos semanais permaneceram expressivos, especialmente nos subsistemas Sul e Sudeste/Centro-Oeste.

<br>**Testes de estacionariedade**

Os testes ADF e KPSS foram aplicados às séries originais e diferenciadas, considerando nível de significância de 5%.

Nas séries originais, os subsistemas Norte e Nordeste apresentaram resultados compatíveis com não estacionariedade em nível. Nos subsistemas Sul e Sudeste/Centro-Oeste, os testes apresentaram resultados divergentes, com rejeição da hipótese de raiz unitária pelo ADF e rejeição da estacionariedade em nível pelo KPSS.

Após a primeira diferenciação, os quatro subsistemas apresentaram rejeição da hipótese de raiz unitária pelo ADF e não rejeição da estacionariedade em nível pelo KPSS.

Esses resultados fornecem evidências favoráveis à utilização de diferenciação não sazonal nos modelos estatísticos, embora não garantam a eliminação de todos os padrões temporais.

<br>**Implicações para a modelagem**

A persistência da autocorrelação semanal indica a necessidade de considerar componentes sazonais com periodicidade de sete dias.

Dessa forma, propõe-se investigar modelos SARIMA com diferenciação não sazonal e componentes autorregressivos e de médias móveis sazonais. A necessidade de diferenciação sazonal adicional será avaliada antes da definição final dos parâmetros.

As alterações metodológicas do ONS em 2021 e 2023 também serão consideradas, pois podem influenciar os testes e a estabilidade dos modelos.

A configuração inicial será definida utilizando apenas os dados de treinamento, evitando utilizar informações do conjunto de teste na seleção dos modelos.
<br><br>

````python
# Comparar a primeira diferença com a diferença sazonal de sete dias e com a combinação das duas

import pandas as pd
import warnings

from statsmodels.tsa.stattools import adfuller, kpss
from statsmodels.tools.sm_exceptions import InterpolationWarning

resultados = []

for sigla, nome in subsistemas.items():

    # Apenas período de treinamento: 2015 a 2022
    serie = (
        df_modelagem[
            (df_modelagem["id_subsistema"] == sigla) &
            (df_modelagem["din_instante"] < "2023-01-01")
        ]
        .sort_values("din_instante")
        .set_index("din_instante")["val_cargaenergiamwmed"]
        .asfreq("D")
    )

    transformacoes = {
        "Original": serie,
        "Primeira diferença (d=1)": serie.diff(),
        "Sazonal semanal (D=1)": serie.diff(7),
        "Combinada (d=1, D=1)": serie.diff().diff(7)
    }

    for tipo, valores in transformacoes.items():
        valores = valores.dropna()

        with warnings.catch_warnings():
            warnings.simplefilter("ignore", InterpolationWarning)

            adf = adfuller(
                valores,
                regression="c",
                autolag="AIC",
                result_object=False
            )

            kpss_result = kpss(
                valores,
                regression="c",
                nlags="auto",
                result_object=False
            )

        resultados.append({
            "Subsistema": sigla,
            "Transformação": tipo,
            "ADF p-valor": adf[1],
            "KPSS p-valor": kpss_result[1],
            "ACF lag 7": valores.autocorr(lag=7),
            "Desvio-padrão": valores.std()
        })

df_diferenciacao = pd.DataFrame(resultados)

print("=== COMPARAÇÃO DAS DIFERENCIAÇÕES ===")
display(df_diferenciacao.round(4))
````


<br>**8.4.2 Comparação das Diferenciações**

<br>Foram avaliadas quatro configurações das séries temporais: original, primeira diferença (d=1), diferenciação sazonal semanal (D=1) e diferenciação combinada (d=1, D=1).

Os testes ADF e KPSS indicaram resultados compatíveis com estacionariedade nas três configurações diferenciadas, considerando nível de significância de 5%.

A primeira diferença reduziu parte da persistência temporal, mas manteve autocorrelação semanal elevada, especialmente nos subsistemas Sul e Sudeste/Centro-Oeste.

A diferenciação sazonal semanal, com período de sete dias, apresentou redução significativa da autocorrelação positiva nessa defasagem. No Norte, o coeficiente passou de 0,894 para -0,357; no Nordeste, de 0,870 para -0,338; no Sul, de 0,824 para -0,339; e no Sudeste/Centro-Oeste, de 0,819 para -0,339.

A diferenciação combinada também apresentou resultados compatíveis com estacionariedade, porém produziu correlações negativas mais intensas na defasagem semanal, indicando a necessidade de investigar possível diferenciação excessiva.

Dessa forma, a configuração com diferenciação sazonal semanal (D=1) será considerada inicialmente na modelagem SARIMA, mantendo a diferenciação combinada como alternativa para comparação.

A seleção definitiva dependerá dos diagnósticos dos resíduos e do desempenho preditivo nos dados de validação, evitando a escolha de parâmetros exclusivamente com base nos testes de estacionariedade.
<br><br>

````python
# Baseline sazonal semanal


import pandas as pd
import numpy as np

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

coluna = "val_cargaenergiamwmed"

resultados_baseline = []
previsoes_baseline = {}

for sigla in ["N", "NE", "S", "SE"]:

    serie = (
        df_modelagem[
            df_modelagem["id_subsistema"] == sigla
        ]
        .sort_values("din_instante")
        .set_index("din_instante")[coluna]
        .asfreq("D")
    )

    # Previsão: mesmo dia da semana anterior
    previsto = serie.shift(7)

    # Período de validação
    validacao = pd.DataFrame({
        "real": serie,
        "previsto": previsto
    }).loc["2023-01-01":"2024-12-31"].dropna()

    mae = mean_absolute_error(
        validacao["real"],
        validacao["previsto"]
    )

    rmse = np.sqrt(mean_squared_error(
        validacao["real"],
        validacao["previsto"]
    ))

    mape = (
        np.mean(
            np.abs(
                (validacao["real"] - validacao["previsto"])
                / validacao["real"]
            )
        ) * 100
    )

    resultados_baseline.append({
        "Subsistema": sigla,
        "MAE": mae,
        "RMSE": rmse,
        "MAPE (%)": mape
    })

    previsoes_baseline[sigla] = validacao

df_baseline = pd.DataFrame(resultados_baseline)

print("=== RESULTADOS DO BASELINE SAZONAL ===")
display(df_baseline.round(2))

````

**9.1 Avaliação do Modelo Baseline Sazonal**

<br>Foi implementado um modelo baseline sazonal com periodicidade semanal de sete dias, utilizando a carga observada na semana anterior como previsão para cada dia.

O modelo foi avaliado no período de validação de 2023 a 2024, utilizando as métricas MAE, RMSE e MAPE.

Os menores erros percentuais foram observados nos subsistemas Norte (2,63%) e Nordeste (3,07%), indicando maior regularidade do comportamento semanal nessas regiões.

O subsistema Sul apresentou o maior MAPE (7,45%), seguido pelo Sudeste/Centro-Oeste (5,05%), demonstrando maior dificuldade de previsão utilizando exclusivamente a observação da semana anterior.

Os resultados confirmam a utilidade do baseline como referência inicial de desempenho, mas também evidenciam suas limitações diante de mudanças de tendência, condições climáticas e variações não representadas pelo padrão semanal.

Esses indicadores serão utilizados para comparar o desempenho dos modelos estatísticos implementados nas próximas etapas, verificando se técnicas mais complexas apresentam ganhos efetivos de precisão.
<br><br>

````python
# SARIMA


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, mean_squared_error

coluna = "val_cargaenergiamwmed"

ordem = (1, 0, 0)
ordem_sazonal = (1, 1, 0, 7)

resultados_sarima = []
previsoes_sarima = {}
modelos_sarima = {}

for sigla in ["N", "NE", "S", "SE"]:

    print(f"\n=== SARIMA: {sigla} ===")

    serie = (
        df_modelagem[
            df_modelagem["id_subsistema"] == sigla
        ]
        .sort_values("din_instante")
        .set_index("din_instante")[coluna]
        .asfreq("D")
    )

    treinamento = serie.loc["2015":"2022"]
    validacao = serie.loc["2023":"2024"]

    # Ajustar parâmetros somente no treinamento
    modelo = SARIMAX(
        treinamento,
        order=ordem,
        seasonal_order=ordem_sazonal,
        trend="n",
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    ajuste = modelo.fit(
        disp=False,
        maxiter=100
    )

    if not ajuste.mle_retvals.get("converged", False):
        print("Aviso: otimização não convergiu.")

    # Aplicar os parâmetros fixos à série com
    # observações disponíveis durante a validação
    serie_ate_validacao = serie.loc["2015":"2024"]

    modelo_validacao = SARIMAX(
        serie_ate_validacao,
        order=ordem,
        seasonal_order=ordem_sazonal,
        trend="n",
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    filtrado = modelo_validacao.filter(ajuste.params)

    # Previsões um passo à frente, sem utilizar
    # observações futuras em cada instante
    previsto = filtrado.get_prediction(
        start=validacao.index[0],
        end=validacao.index[-1],
        dynamic=False
    ).predicted_mean

    comparacao = pd.DataFrame({
        "real": validacao,
        "previsto": previsto
    }).dropna()

    mae = mean_absolute_error(
        comparacao["real"],
        comparacao["previsto"]
    )

    rmse = np.sqrt(mean_squared_error(
        comparacao["real"],
        comparacao["previsto"]
    ))

    mape = np.mean(
        np.abs(
            (comparacao["real"] - comparacao["previsto"])
            / comparacao["real"]
        )
    ) * 100

    resultados_sarima.append({
        "Subsistema": sigla,
        "MAE": mae,
        "RMSE": rmse,
        "MAPE (%)": mape
    })

    modelos_sarima[sigla] = ajuste
    previsoes_sarima[sigla] = comparacao

df_sarima = pd.DataFrame(resultados_sarima)

print("\n=== RESULTADOS DO SARIMA ===")
display(df_sarima.round(2))

# Comparação direta com o baseline
comparativo = df_baseline.merge(
    df_sarima,
    on="Subsistema",
    suffixes=("_Baseline", "_SARIMA")
)

comparativo["Ganho MAPE (%)"] = (
    (comparativo["MAPE (%)_Baseline"]
     - comparativo["MAPE (%)_SARIMA"])
    / comparativo["MAPE (%)_Baseline"] * 100
)

print("\n=== COMPARAÇÃO BASELINE x SARIMA ===")
display(comparativo.round(2))
````


<br>**9.2 Resultados e Avaliação do Modelo SARIMA**

No subsistema Norte, o MAPE foi reduzido de 2,63% para 1,74%, representando uma melhora relativa de 34,02%. No Nordeste, o erro passou de 3,07% para 2,07%, com redução de 32,69%.

Os maiores ganhos relativos ocorreram nos subsistemas Sul e Sudeste/Centro-Oeste, com reduções de 47,47% e 47,96% no MAPE, respectivamente.

Os resultados também demonstram a importância da análise exploratória na escolha da configuração inicial do modelo, especialmente na identificação da sazonalidade semanal.

<br>**9.3 Visualização das Previsões e Análise dos Resíduos**

<br>Após a avaliação quantitativa dos modelos, será realizada uma análise visual das previsões produzidas pelo SARIMA e dos respectivos erros.

Inicialmente, serão comparadas as cargas reais e previstas nos últimos 90 dias de 2024, permitindo verificar a capacidade do modelo de acompanhar as oscilações diárias de cada subsistema.

Em seguida, serão examinados os erros de previsão obtidos durante o período de validação de 2023 a 2024, com o objetivo de identificar possíveis desvios sistemáticos, observações extremas e padrões temporais não representados pelo modelo.

Por fim, serão utilizadas a função de autocorrelação (ACF) e o teste de Ljung-Box para avaliar a presença de dependência temporal remanescente nos erros.

Essa análise permitirá identificar limitações do modelo SARIMA e orientar eventuais refinamentos na etapa final do projeto.
<br><br>

````python

import matplotlib.pyplot as plt

subsistemas = {
    "N": "Norte",
    "NE": "Nordeste",
    "S": "Sul",
    "SE": "Sudeste/Centro-Oeste"
}

fig, axes = plt.subplots(4, 1, figsize=(15, 14))

for ax, (sigla, nome) in zip(axes, subsistemas.items()):

    dados = previsoes_sarima[sigla].tail(90)

    ax.plot(
        dados.index,
        dados["real"],
        label="Carga real",
        linewidth=1.8
    )

    ax.plot(
        dados.index,
        dados["previsto"],
        label="Previsão SARIMA",
        linewidth=1.5,
        linestyle="--"
    )

    ax.set_title(nome)
    ax.set_ylabel("Carga (MWmédio)")
    ax.grid(alpha=0.3)
    ax.legend()

axes[-1].set_xlabel("Data")

fig.suptitle(
    "Carga Real × Previsão SARIMA — Últimos 90 Dias de 2024",
    fontsize=15
)

plt.tight_layout()
plt.show()
````

````python

import matplotlib.pyplot as plt

fig, axes = plt.subplots(4, 1, figsize=(15, 12))

for ax, (sigla, nome) in zip(axes, subsistemas.items()):

    dados = previsoes_sarima[sigla]
    erros = dados["real"] - dados["previsto"]

    ax.plot(
        erros.index,
        erros,
        linewidth=0.7
    )

    ax.axhline(
        0,
        color="black",
        linestyle="--",
        linewidth=1
    )

    ax.set_title(f"Erros de Previsão — {nome}")
    ax.set_ylabel("Erro (MWmédio)")
    ax.grid(alpha=0.3)

axes[-1].set_xlabel("Data")

fig.suptitle(
    "Erros de Previsão SARIMA — Validação 2023–2024",
    fontsize=15
)

plt.tight_layout()
plt.show()
````

````python

import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.stats.diagnostic import acorr_ljungbox

resultados_residuos = []

fig, axes = plt.subplots(4, 1, figsize=(13, 12))

for ax, (sigla, nome) in zip(axes, subsistemas.items()):

    dados = previsoes_sarima[sigla]
    erros = (dados["real"] - dados["previsto"]).dropna()

    # ACF dos erros
    plot_acf(
        erros,
        lags=35,
        ax=ax,
        title=f"ACF dos Erros de Previsão — {nome}"
    )

    ax.grid(alpha=0.2)

    # Teste Ljung-Box
    lb = acorr_ljungbox(
        erros,
        lags=[7, 14, 21, 28],
        return_df=True
    )

    for lag, linha in lb.iterrows():
        resultados_residuos.append({
            "Subsistema": sigla,
            "Lag": lag,
            "LB estatística": linha["lb_stat"],
            "p-valor": linha["lb_pvalue"]
        })

plt.tight_layout()
plt.show()

df_ljungbox = pd.DataFrame(resultados_residuos)

print("=== TESTE DE LJUNG-BOX ===")
display(df_ljungbox.round(5))
````

O modelo acompanhou as principais oscilações diárias e reproduziu os padrões semanais identificados na análise exploratória.

<br>**9.3.2 Análise da Autocorrelação dos Erros**

A análise da função de autocorrelação (ACF) dos erros de previsão demonstrou a presença de correlações remanescentes nos quatro subsistemas.

Foram observados picos negativos nas defasagens semanais de 7 e 14 dias, além de autocorrelação de curto prazo em determinados subsistemas.

A presença dessas correlações sugere a necessidade de investigar configurações alternativas, considerando diferentes combinações de parâmetros autorregressivos, de médias móveis e de diferenciação.

Esses refinamentos serão avaliados com base no desempenho preditivo e nos diagnósticos dos erros, preservando o conjunto de teste de 2025 para a avaliação final.


<br>**9.3.3 Teste de Ljung-Box**

Para complementar a análise dos erros de previsão, foi aplicado o teste de Ljung-Box nas defasagens de 7, 14, 21 e 28 dias, considerando nível de significância de 5%.

O teste avalia a hipótese nula de ausência de autocorrelação conjunta até a defasagem analisada.

Os resultados apresentaram p-valores inferiores a 0,0001 em todos os subsistemas e defasagens avaliadas, permitindo rejeitar a hipótese nula.

Na defasagem de 28 dias, as estatísticas do teste foram de 139,53 para o Norte, 108,34 para o Nordeste, 167,07 para o Sul e 163,17 para o Sudeste/Centro-Oeste.

Esses resultados confirmam a presença de dependência temporal remanescente nos erros de previsão, conforme identificado anteriormente nos gráficos ACF.

Embora o SARIMA(1,0,0)(1,1,0,7) tenha apresentado desempenho superior ao baseline sazonal, a presença de autocorrelação nos erros indica que parte da estrutura temporal das séries ainda não foi completamente representada.

Dessa forma, o modelo implementado será mantido como referência estatística inicial, enquanto configurações alternativas poderão ser investigadas na etapa final do projeto.

Os refinamentos deverão considerar tanto a redução dos erros de previsão quanto a melhoria dos diagnósticos estatísticos, mantendo o ano de 2025 reservado para a avaliação final.
